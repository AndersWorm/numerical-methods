```python
import numpy as np
import matplotlib.pyplot as plt
def lagx(x,xp):
    l_j_data={}
    for xj in xp:
        l_j=1
        for xi in xp:
            if xi==xj:
                continue
            l_j*= (x-xi)/(xj-xi)
        l_j_data[xj]=l_j
    return(l_j_data)
```

```python
xp=np.array([-1,0,1,2])
x=np.linspace(-1,2,)
lagrange=lagx(x,xp)
plt.figure(figsize=(8, 5))
for xj, lj in lagrange.items():
    plt.plot(x, lj, lw=2, label=f"$x_j$ = {xj}")

plt.axhline(0, color="0.7", lw=0.8, zorder=0)
plt.axhline(1, color="0.7", lw=0.8, ls=":", zorder=0)
plt.plot(xp, np.zeros_like(xp), "k|", ms=12)  

plt.xlabel("x"); plt.ylabel(r"$\ell_j(x)$")
plt.legend(); plt.show()
```

```python
def lagx(x,xp):
    l_j_data={}
    for xj in xp:
        l_j=1
        for xi in xp:
            if xi==xj:
                continue
            l_j*= (x-xi)/(xj-xi)
        l_j_data[xj]=l_j
    return(l_j_data)
def lagint(x,xp,yp):
    L=0
    lag=lagx(x,xp) #get all my lagrange functions
    for j, (xj, lj) in enumerate(lag.items()): # enumerate gives me the index j and the xj,lj values
        L += lj * yp[j] # use the index j to get the corresponding y value from yp
    return(L)
```

```python
yp=([2,-1,1,0])
```

```python
lagint(x,xp,yp)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, lagint(x,xp,yp), lw=2, label="Lagrange interpolation")
plt.plot(xp, yp, "ro", label="Data points")
plt.axhline(0, color="0.7", lw=0.8, zorder=0)
plt.axhline(1, color="0.7", lw=0.8, ls=":", zorder=0)
plt.xlabel("x"); plt.ylabel("L(x)")
plt.legend(); plt.show()
```
