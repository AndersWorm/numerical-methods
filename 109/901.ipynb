```python
def zerof(A,m):
    B= A.copy()
    for i in range(B.shape[0]):
        for j in range(B.shape[1]):
            if abs(A[i,j])<m:
                 B[i,j]=0
    return(A)
```

```python
from scipy import sparse as sp
from scipy.sparse import linalg as sla
import numpy as np
A=(np.random.randn(20,20)-0.5)**5 # randn is the normal distribution with mean value zero and standard deviation 1 unlike rand which is uniformly values between 0 and 1
# ^5 works on every index and -1/2 subtracts from every index
def zerof(A,m):
    B=A.copy()
    B[abs(A)<m]=0 #abs(A) runs through the entire matrix and compares them to m
    #and returns a true and false array which A[] uses and replace all the true with 0
    return(B)
print(zerof(A,0.1))
```

```python
from scipy import sparse as sp
from scipy.sparse import linalg as sla
import numpy as np
A=sp.random(5000,5000,density=0.1,format="csc") #sparse matrix where 10% of the numbers are different from 0
B=sp.random(5000,1,density=0.1,format="csc")
x=sla.spsolve(A,B) #sparse version of la.solve returns normal array
c=A.todense()@x - B.todense().flatten()
```

```python
print("c=",np.linalg.norm(c,2))
```
